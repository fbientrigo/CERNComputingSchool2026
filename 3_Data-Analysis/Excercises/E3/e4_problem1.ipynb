# Problem 1

- 1% of events are product of this cscparticle with mass $m_{scs}$
- the spectrum its $m_{e\bar e}$
- the pdf signal its a gaussian distribution $\sigma \approx 0.2 \, GeV$
- the particle its in the range of $20 GeV < m_{scs} < 40 GeV$

```python
import matplotlib.pyplot as plt
import numpy as np
import scipy.integrate as integrate
import random
from scipy.optimize import curve_fit
from scipy.special import erfcinv
import scipy.stats as st

# Problem 1
with open('LHC_data_2023.txt', 'r') as f:
    lines = (line.strip() for line in f if line)
    real_data = [float(line) for line in lines]

print(f"This file containes {len(real_data)} entries that range from {min(real_data)} Gev to {max(real_data)} Gev.")

lower_bound = min(real_data)
upper_bound = max(real_data)

def pdf_theory(x, mass_csc):
    array = (np.exp(-(x-mass_csc)**2))
    # normalize this array
    norm_const = integrate.quad(lambda t: np.exp(-(t-mass_csc)**2), 0, 100)[0]
    return array / norm_const


x = np.linspace(lower_bound,upper_bound,100)
plt.plot(x, pdf_theory(x, 25))
```
